```python
import numpy as np
from math import log2

def analyze_bitstream(counts, label=""):
    zeros = counts.get('0', 0)
    ones = counts.get('1', 0)
    total = zeros + ones
    
    p0 = zeros / total
    p1 = ones / total
    
    bias_pct = abs(p1 - 0.5) * 100
    
    p_max = max(p0, p1)
    min_entropy = -log2(p_max)
    
    print(f"--- {label} ---")
    print(f"Total shots: {total}")
    print(f"P(0) = {p0:.4f}, P(1) = {p1:.4f}")
    print(f"Bias from ideal 50/50: {bias_pct:.2f}%")
    print(f"Min-entropy: {min_entropy:.4f} bits (ideal = 1.0)")
    print()
    
    return {"p0": p0, "p1": p1, "bias_pct": bias_pct, "min_entropy": min_entropy}
```

```python
def von_neumann_extractor(bits):
    """
    Standard von Neumann extractor.
    Takes bits in pairs: 01 -> 0, 10 -> 1, 00/11 -> discarded
    """
    output = []
    i = 0
    while i < len(bits) - 1:
        pair = bits[i] + bits[i+1]
        if pair == '01':
            output.append('0')
        elif pair == '10':
            output.append('1')
        # 00 and 11 are discarded, no output bit
        i += 2
    return output

vn_output = von_neumann_extractor(raw_bits_q0)
print(f"Input bits: {len(raw_bits_q0)}")
print(f"Von Neumann output bits: {len(vn_output)}")
print(f"Efficiency: {len(vn_output)/len(raw_bits_q0)*100:.1f}%")
```

```python
def bias_aware_extractor(bits, window_size=3):
    """
    Custom extractor tailored to kingston qubit 0's bias.
    XORs groups of 'window_size' consecutive bits together.
    XOR-ing biased bits together pushes the result closer to 50/50,
    and unlike von Neumann, we don't discard any bits outright.
    """
    output = []
    for i in range(0, len(bits) - window_size + 1, window_size):
        chunk = bits[i:i+window_size]
        xor_result = 0
        for bit in chunk:
            xor_result ^= int(bit)
        output.append(str(xor_result))
    return output

custom_output = bias_aware_extractor(raw_bits_q0, window_size=3)
print(f"Input bits: {len(raw_bits_q0)}")
print(f"Custom extractor output bits: {len(custom_output)}")
print(f"Efficiency: {len(custom_output)/len(raw_bits_q0)*100:.1f}%")

# Check how biased the OUTPUT is now
ones = custom_output.count('1')
zeros = custom_output.count('0')
print(f"Output P(0)={zeros/len(custom_output):.4f}, P(1)={ones/len(custom_output):.4f}")
```

```python
import numpy as np
from scipy import stats
import math
def toeplitz_extractor(bits, m, seed=42):
    """
    Universal-hash (Toeplitz matrix) randomness extractor.
    Multiplies the raw bit vector by a random Toeplitz matrix mod 2.
    By the Leftover Hash Lemma: if m <= n*min_entropy - security_margin,
    the output is provably close to uniform -- regardless of the specific
    bias/correlation structure of the raw input (unlike a fixed-window XOR).
    """
    bits = np.array([int(b) for b in bits], dtype=np.int64)
    n = len(bits)
    rng = np.random.default_rng(seed)
    toeplitz_seed = rng.integers(0, 2, size=n + m - 1, dtype=np.int64)
    T = np.zeros((m, n), dtype=np.int64)
    for i in range(m):
        T[i] = toeplitz_seed[m - 1 - i: m - 1 - i + n]
    output_bits = (T @ bits) % 2
    return [str(b) for b in output_bits]

# --- Estimate min-entropy directly from the raw Kingston Q0 stream ---
raw_ints = [int(b) for b in raw_bits_q0]
p1_hat = sum(raw_ints) / len(raw_ints)
min_entropy_per_bit = -math.log2(max(p1_hat, 1 - p1_hat))
n = len(raw_ints)
print(f"Estimated min-entropy per raw bit: {min_entropy_per_bit:.4f}")
print(f"Total extractable entropy budget: {n * min_entropy_per_bit:.1f} bits\n")

# --- Extract, leaving a small security margin below the theoretical max ---
security_margin = 64  # bits
m = int(n * min_entropy_per_bit) - security_margin
toeplitz_output = toeplitz_extractor(raw_bits_q0, m, seed=42)

print(f"Input bits: {n}")
print(f"Toeplitz extractor output bits: {len(toeplitz_output)}")
print(f"Efficiency: {len(toeplitz_output)/n*100:.1f}%")
toeplitz_counts = {'0': toeplitz_output.count('0'), '1': toeplitz_output.count('1')}
print(f"Output P(0)={toeplitz_counts['0']/len(toeplitz_output):.4f}, "
      f"P(1)={toeplitz_counts['1']/len(toeplitz_output):.4f}\n")

# --- Reuse your existing analyze_bitstream() for consistent reporting ---
print("Toeplitz-Hash Extractor Output:")
toeplitz_stats = analyze_bitstream(toeplitz_counts, "Toeplitz Extractor Output")

# --- Chi-square uniformity test, added on top of your existing bias/entropy metrics ---
def chi_square_uniformity(counts, label=""):
    zeros, ones = counts.get('0', 0), counts.get('1', 0)
    total = zeros + ones
    expected = total / 2
    chi2 = ((zeros - expected) ** 2 + (ones - expected) ** 2) / expected
    pval = 1 - stats.chi2.cdf(chi2, df=1)
    verdict = "PASS (uniform)" if pval > 0.01 else "FAIL (biased)"
    print(f"{label}: chi2={chi2:.4f}, p-value={pval:.4f} -> {verdict} at alpha=0.01")
    return chi2, pval

print("=== Chi-square uniformity test (all three extractors) ===")
vn_counts = {'0': vn_output.count('0'), '1': vn_output.count('1')}
custom_counts = {'0': custom_output.count('0'), '1': custom_output.count('1')}
chi_square_uniformity(vn_counts, "Von Neumann")
chi_square_uniformity(custom_counts, "Custom XOR (k=3)")
chi_square_uniformity(toeplitz_counts, "Toeplitz-hash")

print("\n=== Efficiency summary ===")
print(f"Von Neumann:   {len(vn_output)}/{n} = {len(vn_output)/n*100:.1f}%")
print(f"Custom XOR:    {len(custom_output)}/{n} = {len(custom_output)/n*100:.1f}%")
print(f"Toeplitz-hash: {len(toeplitz_output)}/{n} = {len(toeplitz_output)/n*100:.1f}%  "
      f"({len(toeplitz_output)/len(vn_output):.2f}x more bits than Von Neumann)")
```

```python
import hashlib

def sha256_extractor(bits, block_bits, output_bits_per_block=256):
    """
    Cryptographic conditioning extractor (the approach used by most
    real-world CSPRNGs, e.g. /dev/urandom-style conditioning).
    Splits the raw bitstream into fixed-size blocks, hashes each block
    with SHA-256, and keeps output_bits_per_block bits of the digest.

    Unlike Toeplitz/Trevisan, this is NOT provably close to uniform via
    the Leftover Hash Lemma -- its security instead relies on SHA-256's
    cryptographic hardness (collision/preimage resistance). Worth stating
    explicitly in your report as a different *class* of guarantee.
    """
    bits = np.array([int(b) for b in bits], dtype=np.uint8)
    n = len(bits)
    output = []
    for start in range(0, n - block_bits + 1, block_bits):
        block = bits[start:start + block_bits]
        packed = np.packbits(block)
        digest = hashlib.sha256(packed.tobytes()).digest()
        digest_bits = np.unpackbits(np.frombuffer(digest, dtype=np.uint8))
        output.extend(digest_bits[:output_bits_per_block].tolist())
    return [str(b) for b in output]

# --- Size the block so each block's entropy budget just covers one 256-bit digest ---
# (same min_entropy_per_bit computed earlier from raw_bits_q0)
block_bits = math.ceil(256 / min_entropy_per_bit)
print(f"Block size: {block_bits} raw bits per SHA-256 block "
      f"(covers a 256-bit digest at {min_entropy_per_bit:.4f} bits entropy/bit)")

sha_output = sha256_extractor(raw_bits_q0, block_bits=block_bits, output_bits_per_block=256)

print(f"\nInput bits: {n}")
print(f"SHA-256 extractor output bits: {len(sha_output)}")
print(f"Efficiency: {len(sha_output)/n*100:.1f}%")
sha_counts = {'0': sha_output.count('0'), '1': sha_output.count('1')}
print(f"Output P(0)={sha_counts['0']/len(sha_output):.4f}, "
      f"P(1)={sha_counts['1']/len(sha_output):.4f}\n")

print("SHA-256 Extractor Output:")
sha_stats = analyze_bitstream(sha_counts, "SHA-256 Extractor Output")

print("=== Chi-square uniformity test (all four extractors) ===")
chi_square_uniformity(vn_counts, "Von Neumann")
chi_square_uniformity(custom_counts, "Custom XOR (k=3)")
chi_square_uniformity(toeplitz_counts, "Toeplitz-hash")
chi_square_uniformity(sha_counts, "SHA-256 block extractor")

print("\n=== Efficiency summary (all four) ===")
print(f"Von Neumann:   {len(vn_output)}/{n} = {len(vn_output)/n*100:.1f}%")
print(f"Custom XOR:    {len(custom_output)}/{n} = {len(custom_output)/n*100:.1f}%")
print(f"Toeplitz-hash: {len(toeplitz_output)}/{n} = {len(toeplitz_output)/n*100:.1f}%")
print(f"SHA-256:       {len(sha_output)}/{n} = {len(sha_output)/n*100:.1f}%")
```

```python
def analyze_bell_pair(counts, label=""):
    """
    Analyzes a 2-qubit Bell-pair measurement result.
    Reports both the entanglement correlation quality AND the
    marginal statistics of qubit 0, treated as the "random bit"
    -- directly comparable to your single-qubit analyze_bitstream().
    """
    total = sum(counts.values())
    p00 = counts.get('00', 0) / total
    p01 = counts.get('01', 0) / total
    p10 = counts.get('10', 0) / total
    p11 = counts.get('11', 0) / total
 
    correlated = p00 + p11          # ideal = 1.0 for a perfect Bell state
    anticorrelated = p01 + p10      # ideal = 0.0 -- this is your error/decoherence rate
 
    # Marginal distribution of qubit 0 (leftmost bit) -- this is the "random bit"
    p0_marginal = p00 + p01
    p1_marginal = p10 + p11
    marginal_bias = abs(p1_marginal - 0.5) * 100
    marginal_entropy = -math.log2(max(p0_marginal, p1_marginal))
 
    print(f"--- {label} ---")
    print(f"Total shots: {total}")
    print(f"P(00)={p00:.4f}  P(01)={p01:.4f}  P(10)={p10:.4f}  P(11)={p11:.4f}")
    print(f"Correlated outcomes (00+11): {correlated*100:.2f}% (ideal 100%)")
    print(f"Anti-correlated / error rate (01+10): {anticorrelated*100:.2f}% (ideal 0%)")
    print(f"Qubit-0 marginal: P(0)={p0_marginal:.4f}, P(1)={p1_marginal:.4f}")
    print(f"Qubit-0 marginal bias from 50/50: {marginal_bias:.2f}%")
    print(f"Qubit-0 marginal min-entropy: {marginal_entropy:.4f} bits (ideal = 1.0)")
    print()
    return dict(p00=p00, p01=p01, p10=p10, p11=p11, correlated=correlated,
                anticorrelated=anticorrelated, marginal_bias=marginal_bias,
                marginal_entropy=marginal_entropy)
 
bell_sim_stats = analyze_bell_pair(counts_bell_sim, "Bell Pair - Simulator")
 
 
```
